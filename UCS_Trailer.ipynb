## 1. Introduction and Problem Definition

In Practical Exercise 3 (Part A) we modelled a self-driving trailer travelling across the Kenyan highway network as a **directed, weighted graph**. Towns are nodes, road segments are directed edges and the edge weights are road distances in kilometres taken from Google Maps.

In this exercise (Part B) the same graph is used. **Uniform Cost Search (UCS)** is run to find the cheapest route from Nairobi to Kisumu, showing the traversal step by step.

### Problem formulation

| Component | Definition |
|---|---|
| **Initial state** | The trailer is in **Nairobi** |
| **Goal state** | The trailer is in **Kisumu** |
| **State space** | The 13 towns in the graph (Nairobi, Ruiru, Thika, Nyeri, Naivasha, Gilgil, Nyahururu, Nakuru, Kericho, Eldoret, Webuye, Kakamega, Kisumu) |
| **Actions** | Drive along an outgoing road from the current town to a neighbouring town |
| **Transition model** | Driving from town *A* along the edge *A to B* puts the trailer in town *B* |
| **Step cost** | The road distance of the edge in km  |
| **Path cost** *g(n)* | The sum of the step costs from Nairobi to node *n* |
| **Goal test** | Applied when a node is **expanded ** from the frontier. This is what ensures optimality. |
| **Solution** | An optimal path from Nairobi to Kisumu. |



## 2. Graph Data Structure

The road network is stored as a Python **dictionary**. Each town lists the towns you can drive to from it, and how far each one is.```

For example, `graph["Nairobi"]["Ruiru"]` is the Nairobi to Ruiru road. Roads only go one way.

The distances come from our Part 1 table (Google Maps).

````python
graph = {
    "Nairobi":   {"Ruiru": 24.6, "Naivasha": 91},
    "Ruiru":     {"Thika": 21.4},
    "Thika":     {"Nyeri": 113},
    "Nyeri":     {"Nyahururu": 104},
    "Naivasha":  {"Gilgil": 30.4},
    "Gilgil":    {"Nyahururu": 67, "Nakuru": 42.6},
    "Nyahururu": {"Nakuru": 63.3},
    "Nakuru":    {"Kericho": 107, "Eldoret": 156},
    "Kericho":   {"Kisumu": 84},
    "Eldoret":   {"Webuye": 71.6},
    "Webuye":    {"Kakamega": 48.5},
    "Kakamega":  {"Kisumu": 53.4},
    "Kisumu":    {},   # goal: no outgoing roads needed
}

START = "Nairobi"
GOAL = "Kisumu"
````

### 2.1 Helper functions (cost function and successors)

These give the search algorithm a clean interface to the graph: the **successor function**  and the **step cost function**.

````python
def successors(town):
    """Return the list of (neighbour, step_cost) pairs reachable from `town`."""
    return list(graph[town].items())


def step_cost(from_town, to_town):
    """Cost (km) of driving directly from `from_town` to `to_town`."""
    return graph[from_town][to_town]


def path_cost(path):
    """Total cost g(n) of a path given as a list of towns."""
    return round(sum(step_cost(a, b) for a, b in zip(path, path[1:])), 1)
````

## 2.2 Quick Check of the Graph

We checked the graph against the Part 1 table and confirmed that our hand-calculated route (Nairobi → Naivasha → Gilgil → Nakuru → Kericho → Kisumu) costs **355.0 km**.

````python
# Part 1 table, row by row (From, To, km)
part1_table = [
    ("Nairobi", "Ruiru", 24.6),     ("Ruiru", "Thika", 21.4),
    ("Thika", "Nyeri", 113),        ("Nyeri", "Nyahururu", 104),
    ("Nyahururu", "Nakuru", 63.3),  ("Nairobi", "Naivasha", 91),
    ("Naivasha", "Gilgil", 30.4),   ("Gilgil", "Nyahururu", 67),
    ("Gilgil", "Nakuru", 42.6),     ("Nakuru", "Kericho", 107),
    ("Kericho", "Kisumu", 84),      ("Nakuru", "Eldoret", 156),
    ("Eldoret", "Webuye", 71.6),    ("Webuye", "Kakamega", 48.5),
    ("Kakamega", "Kisumu", 53.4),
]

# 1. The graph has exactly the edges (and directions) in the table
graph_edges = {(a, b, w) for a, nbrs in graph.items() for b, w in nbrs.items()}
assert graph_edges == set(part1_table), "Graph does not match the Part 1 table"

# 2. Kisumu can be reached from Nairobi
def reachable(start):
    seen, stack = set(), [start]
    while stack:
        t = stack.pop()
        if t not in seen:
            seen.add(t)
            stack.extend(graph[t])
    return seen

assert GOAL in reachable(START), "Kisumu is not reachable from Nairobi"

# 3. Hand-calculated route from Part 1
route = ["Nairobi", "Naivasha", "Gilgil", "Nakuru", "Kericho", "Kisumu"]
assert path_cost(route) == 355.0

print(f"Graph OK: {len(graph)} towns, {len(graph_edges)} roads, matches the Part 1 table.")
print("Part 1 route cost:", path_cost(route), "km")
````
